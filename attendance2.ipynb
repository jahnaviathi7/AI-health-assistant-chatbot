```python
# Install packages
!pip install pandas openpyxl scikit-learn -q

import pandas as pd
import glob
from google.colab import files
from sklearn.tree import DecisionTreeClassifier

# =====================================
# 1. Upload CSV Files
# =====================================
print("Upload both CSV files")
uploaded = files.upload()

# =====================================
# 2. Read and Merge Data
# =====================================
csv_files = glob.glob("*.csv")

print("\nUploaded Files:")
for f in csv_files:
    print(f)

df = pd.concat(
    [pd.read_csv(file) for file in csv_files],
    ignore_index=True
)

print("\nMerged Dataset Shape:")
print(df.shape)

print("\nFirst 5 Rows:")
display(df.head())

# =====================================
# 3. Show Column Names
# =====================================
print("\nColumns:")
print(df.columns.tolist())

# =====================================
# 4. Attendance Calculation
# =====================================
total_meeting_minutes = (
    df.groupby("ID")["Duration (minutes)"]
      .first()
      .sum()
)

print("\nTotal Meeting Minutes:")
print(total_meeting_minutes)

attendance = (
    df.groupby(
        ["Email", "Name (original name)"]
    )["Duration (minutes).1"]
    .sum()
    .reset_index()
)

attendance["Attendance_Percentage"] = (
    attendance["Duration (minutes).1"]
    / total_meeting_minutes
) * 100

attendance["Attendance_Percentage"] = (
    attendance["Attendance_Percentage"]
    .round(2)
)

attendance["Eligibility"] = attendance[
    "Attendance_Percentage"
].apply(
    lambda x:
    "Eligible"
    if x >= 80
    else "Not Eligible"
)

# =====================================
# 5. Display Results
# =====================================
print("\nAttendance Report:")
display(attendance.head(20))

# =====================================
# 6. Eligibility Summary
# =====================================
eligible_count = (
    attendance["Eligibility"] == "Eligible"
).sum()

not_eligible_count = (
    attendance["Eligibility"] == "Not Eligible"
).sum()

print("\nEligibility Summary")
print("Eligible Candidates:", eligible_count)
print("Not Eligible Candidates:", not_eligible_count)

# =====================================
# 7. Machine Learning Model
# =====================================
attendance["Target"] = attendance[
    "Eligibility"
].map({
    "Eligible": 1,
    "Not Eligible": 0
})

X = attendance[["Attendance_Percentage"]]
y = attendance["Target"]

model = DecisionTreeClassifier(random_state=42)

model.fit(X, y)

print("\nML Model Trained Successfully")

# =====================================
# 8. Sample Prediction
# =====================================
test_values = [[75], [82], [95]]

predictions = model.predict(test_values)

print("\nPrediction Results")

for value, pred in zip(test_values, predictions):
    status = "Eligible" if pred == 1 else "Not Eligible"
    print(f"Attendance {value[0]}% --> {status}")

# =====================================
# 9. Save Excel File
# =====================================
output = attendance[
    [
        "Name (original name)",
        "Email",
        "Attendance_Percentage",
        "Eligibility"
    ]
]

excel_file = "Candidate_Eligibility_Report.xlsx"

output.to_excel(
    excel_file,
    index=False
)

print("\nExcel File Created:")
print(excel_file)

# =====================================
# 10. Show Final Output
# =====================================
print("\nFinal Report Preview:")
display(output.head(20))

# =====================================
# 11. Download Excel
# =====================================
files.download(excel_file)
```
